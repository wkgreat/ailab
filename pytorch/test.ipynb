```python
dx, dy = torch.meshgrid(torch.arange(4), torch.arange(5), indexing="xy")
dx, dy
```

```python
coords = torch.stack([dy, dx], dim=2).reshape(-1, 2)
coords
```

```python
splats[:, None, :].shape, coords[None, :, :].shape
torch.cat(
    [
        splats[:, None, :].expand(splats.size(0), coords.size(0), splats.size(1)),
        coords[None, :, :].expand(splats.size(0), coords.size(0), coords.size(1)),
    ],
    dim=-1,
)
```

```python
features = torch.randn(6, 3)
print("features: \n", features)
group_id = torch.tensor([0, 1, 0, 2, 1, 2])
print("group_id: \n", group_id)

out = torch.zeros(3, 3)
print("out: \n", out)

out.scatter_add_(dim=0, index=group_id[:, None].expand_as(features), src=features)
print("index: \n", group_id[:, None].expand_as(features))

print(out)
```
