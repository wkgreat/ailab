## 自动微分

```python
x = torch.arange(4.0)
x.requires_grad_(True)
x, x.grad
```

```python
y = 2 * torch.dot(x,x)
y
```

```python
y.backward()
x.grad
```

```python
x.grad.zero_()

y = x.sum()
y.backward()
y, x, x.grad
```
